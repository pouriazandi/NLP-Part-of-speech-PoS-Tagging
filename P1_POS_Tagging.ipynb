```python
def read_conllu(filename):
    words = []
    tags = []
    current_word= []
    current_tag = []

    with open(filename, "r", encoding="utf-8") as sentences:
        
        for line in sentences:

            # Ignore comments
            if line.startswith("#"):
                continue

            if line.strip() != "":
                fields= line.split("\t")
                if "-" in fields[0]:
                    continue
                elif "." in fields[0]:
                    continue
                else:
                    current_word.append(fields[1])
                    current_tag.append(fields[3])
            elif line.strip() == "":
                if len(current_word) <= 128:
                    words.append(current_word)
                    tags.append(current_tag)
                current_word = []
                current_tag = []
                
        if current_word and len(current_word) <= 128:
            words.append(current_word)
            tags.append(current_tag)
    return words, tags      
```

```python
words, tags = read_conllu("1-data/1-English/en_ewt-ud-train.conllu")
print(words[0])
print(tags[0])
print(len(words))
print(len(tags))
print(len(words[0]))
print(len(tags[0]))

print(max(len(sentence) for sentence in words))
```
